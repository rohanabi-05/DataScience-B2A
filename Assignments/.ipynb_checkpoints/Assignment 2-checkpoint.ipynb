#### Q1. employee = ("Ali", 25, "Data Analyst", 75000)

#### Q2. You are given a list that contains some duplicate items. 
Remove the duplicates by converting the list to a set. Final output should be in list

shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

Expected Output: ['apple', 'grape', 'banana', 'orange']

```python
shopping = ["apple", "banana", "apple", "orange", "banana", "grape"]
```

```python
# Converting list into set
shopping_list = set(shopping)
print(shopping_list)
```

```python
# Converting back to List
print(list(shopping_list))
```

#### Q3. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the student enrollment. 
Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in "Course A" but not in "Course B" (difference). Find students who are only in one of the two courses (symmetric difference). course_a = {"John", "Alice", "Bob", "David"} course_b = {"Alice", "Eve", "Charlie", "David"}

Using set methods

• Find students who are in Course A but not in Course B (difference)

• Find students who are only in one of the two courses (symmetric difference)

Expected Output:

Students in Course A but not in Course B: {'Bob', 'John'}

Students only in one of the two courses: {'Eve', 'John', 'Bob', 'Charlie'}

```python
course_a = {"John", "Alice", "Bob", "David"} 
course_b = {"Alice", "Eve", "Charlie", "David"}
```

```python
# Finding names in both courses
course_a - course_b
```

```python
# Finding names only in 1 course
course_a ^ course_b
```

#### Q6. Set Comparison
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

Find common elements.
Find elements only in Set A.
Find elements only in Set B.

```python
A = {1, 2, 3, 4, 5} 
B = {4, 5, 6, 7}

# Finding common elements
print(A & B)
```

```python
# Finding numbers only in A
print(A - B)
```

```python
# Finding number only in B 
print(B - A)
```
