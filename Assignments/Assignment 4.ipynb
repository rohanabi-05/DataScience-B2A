#### Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
user_no = 0
```

```python
for email in users:
    if user_no == 3:
        print("Limit reached. Stop sending emails.")
        break
        
    print("Email sent to", email)
    user_no += 1   
```

#### Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] 
critical = 40
```

```python
for t in temps:
    if t > 40:
        print("\n Critical temperature reached! Stop monitoring.")
        break
        
    print("\n Temperature : ", t)
```

#### Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
total = 0
```

```python
for price in cart:
    total += price

print("Total Price:", total)
```

#### Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True}, {"name": "Sara", "paid": False}, {"name": "John", "paid": True} ]
```

```python
for c in customers:
    if not c["paid"]:
        continue
        
    print("Access granted to : ", c['name'])
```

#### Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
dom = []
```

```python
for mail in emails:
    dom.append(mail.split("@")[1])

print("Domains:", dom)
```

#### Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
```

```python
for i in numbers:
    for j in multipliers: 
        print(f" {i} x {j} = {i * j}")
```

#### Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "python123"
password = input("Enter you password: ")
```

```python
while password != correct_password:
    print("Incorrect password. Try again.")
    password = input("Enter you password: ")

print("Access granted!")
```

#### Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
numbers = list(range(5, 0, -1))

while len(numbers) > 0:
    num = numbers.pop(0)
    print("\n", num)
print("\n Time's up!")
```
