```python
class student:
    def __init__ (self, school_name, location):
        self.sclname = school_name
        self.loc = location
        print("welcome!!")
    
    def greet(self,name,age):
        self.Name = name
        self.Age = age
        print(f'My name is {self.Name}, and my age is {self.Age}')

    def sub(self):
        return 'we are learnin oops'
```

```python
a = student('Rooh', 21)
```

```python
a.greet('Rooh', 21)
```
